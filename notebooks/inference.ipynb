```python
import sys
sys.path.insert(0, '..')
# from speaker_classifier_modules.model import VoiceClassifier
from speaker_classifier_modules.model import VoiceClassifier
from speaker_classifier_modules.data import transform
```

```python
model = VoiceClassifier()
model.load_state_dict(torch.load(r"D:\audio_classifier\models\best_model_20260812_192422.pth"))
model.to(device=torch.device('cuda' if torch.cuda.is_available() else 'cpu'))
```

```python
# inference
class_dict = {0: 'ansh', 1: 'varsha'}
# step 1 is to load the model and the state dict
# model = ImgTxtClassifier()
# model.load_state_dict(torch.load("ansh_varsha_classifier_9336.pth", map_location=torch.device('cpu')))

# step 2 is to set the model to evaluation mode
model.eval()

# path of the image to be classified
from PIL import Image
path = r"D:\audio_classifier\temp_spectrogram.png"

# step 3 is to load the image and apply the same transformations as during training
image = Image.open(path).convert("RGB") #this is converted to RGB to prevent errors when the image is in grayscale
transformed_image = transform(image).unsqueeze(0)  # add batch dimension

# step 4 is to pass the image through the model and get the predicted class
with torch.no_grad():
    transformed_image = transformed_image.to(torch.device("cuda" if torch.cuda.is_available() else "cpu"))
    output = model(transformed_image)
    _, predicted_class = torch.max(output.data, 1)
    print("Predicted class:", predicted_class.item())
    print("Predicted class name:", class_dict[predicted_class.item()])
```
